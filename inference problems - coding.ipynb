### Question 7

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 
import seaborn as sns 
```

```python
df7 = pd.read_csv('./data/ames_prices.csv')
```

```python
# add indicator
df7['Binary_CA'] = np.where(df7['Central.Air'] == 'Y', 1, 0)

# estimate prop of houses with CA
sum(df7['Binary_CA'])/len(df7)
```

```python
# bootstrap estimates of proportion

# orignial proportion
p_hat = sum(df7['Binary_CA'])/len(df7)

# Bootstrap settings
B = 2000
n = len(df7)

rng = np.random.default_rng(123)

# Bootstrap resampling
bootstrap_props = np.empty(B)

for i in range(B):
    sample = df7["Binary_CA"].sample(
        n=n,
        replace=True,
        random_state=int(rng.integers(0, 2**32))
    )

    bootstrap_props[i] = sample.mean()

print("Sample proportion:", p_hat)
print("Bootstrap standard error:", bootstrap_props.std(ddof=1))
```

```python
bootstrap_se = np.std(bootstrap_props, ddof=1)
from scipy.stats import norm

# confidence levels and z values
for confidence in [0.90, 0.95, 0.99]:
    z_star = norm.ppf(1 - (1 - confidence) / 2)

    lower = p_hat - z_star * bootstrap_se
    upper = p_hat + z_star * bootstrap_se

    print(f"{confidence:.0%} CI: ({lower:.4f}, {upper:.4f})")
```

Based on these results, if the null hypothesis H0 is that p = 90, we reject this null at all three confidence levels tested and conclude that H0 is not .9 based on these conditions. .9 is not in any of the confidence intervals because the standard error is very small. 

### Question 8

```python
correlation = np.arcsinh(df7['area']).corr(np.arcsinh(df7['price']))
print("Pearson correlation:", correlation)
```

```python
# original stat
p_hat = correlation

B = 2000
n = len(df7)
rng = np.random.default_rng(123)

bootstrap_corrs = np.empty(B)

for i in range(B):
    # sample rows
    idx = rng.integers(0, n, size=n)

    # resample columns using the same indices
    sample_x = df7["area"].iloc[idx]
    sample_y = df7["price"].iloc[idx]

    # Calculate correlation
    bootstrap_corrs[i] = np.arcsinh(sample_x).corr(np.arcsinh(sample_y))



# Bootstrap standard error
bootstrap_se = np.std(bootstrap_corrs, ddof=1)

print("Original correlation:", p_hat)
print("Bootstrap standard error:", bootstrap_se)
```

```python
bootstrap_se = np.std(bootstrap_corrs, ddof=1)
from scipy.stats import norm

# confidence levels and z values
for confidence in [0.90, 0.95, 0.99]:
    z_star = norm.ppf(1 - (1 - confidence) / 2)

    lower = p_hat - z_star * bootstrap_se
    upper = p_hat + z_star * bootstrap_se

    print(f"{confidence:.0%} CI: ({lower:.4f}, {upper:.4f})")
```

Based on these results, if the null hypothesis H0 is that p = 0, we reject this null at all three confidence levels tested and conclude that H0 is not 0 based on these conditions. 0 is not in any of the confidence intervals. This means that there is likely an association between area and price. In this case, because it is positive and appears to be somewhere between .7 and .75, the two columns are strongly positively correlated with each other. This means that houses that have larger area will mostly likely be pricier. This makes sense intuitively, because while other things like location matter, larger houses are generally more expensive.

Area and Price must remain grouped in the resampling process because each area has a corresponding price, and we want to understand their relationship. If we don't keep them paired, then random values will be drawn from each column that will have no relationship with each other, and it will fail to evaluate what we are looking for.

```python
df9 = pd.read_csv("./data/metabric.csv")
```

```python
mean = df9['Overall Survival (Months)'].mean()
mean
```

```python
# original stat
p_hat = mean

B = 2000
n = len(df9)
rng = np.random.default_rng(123)

bootstrap_means = np.empty(B)

for i in range(B):
    sample = df9["Overall Survival (Months)"].sample(
        n=n,
        replace=True,
        random_state=int(rng.integers(0, 2**32))
    )

    bootstrap_means[i] = sample.mean()

print("Sample proportion:", p_hat)
print("Bootstrap standard error:", bootstrap_means.std(ddof=1))
```

```python
bootstrap_se = np.std(bootstrap_means, ddof=1)
from scipy.stats import norm

# confidence levels and z values
for confidence in [0.90, 0.95, 0.99]:
    z_star = norm.ppf(1 - (1 - confidence) / 2)

    lower = p_hat - z_star * bootstrap_se
    upper = p_hat + z_star * bootstrap_se

    print(f"{confidence:.0%} CI: ({lower:.4f}, {upper:.4f})")
```

Based on these results, if the null hypothesis H0 is that p = 120, we reject this null at all three confidence levels tested and conclude that H0 is not 0 based on these conditions. 120 is not in any of the confidence intervals. 

The results do not estimate unrestricted mean lifetime because living patients' survival times are right-censored which means their recorded times indicate how long they were followed, not how long they ultimately lived. Their eventual death times are unknown and may be much later than their last follow-up.

```python
df10 = pd.read_csv("./data/nhanes_data_17_18.csv")
```

```python
df10 = df10.loc[(df10['AgeInYearsAtScreening'] >= 18) & (df10["SleepHoursWeekdaysOrWorkdays"] != "Nan")]
df10.head()
```

```python
mean = df10['SleepHoursWeekdaysOrWorkdays'].mean()
```

```python
# original stat
p_hat = mean

B = 2000
n = len(df10)
rng = np.random.default_rng(123)

bootstrap_means = np.empty(B)

for i in range(B):
    sample = df10["SleepHoursWeekdaysOrWorkdays"].sample(
        n=n,
        replace=True,
        random_state=int(rng.integers(0, 2**32))
    )

    bootstrap_means[i] = sample.mean()

print("Sample proportion:", p_hat)
print("Bootstrap standard error:", bootstrap_means.std(ddof=1))
```

```python
bootstrap_se = np.std(bootstrap_means, ddof=1)


# confidence levels and z values
for confidence in [0.90, 0.95, 0.99]:
    z_star = norm.ppf(1 - (1 - confidence) / 2)

    lower = p_hat - z_star * bootstrap_se
    upper = p_hat + z_star * bootstrap_se

    print(f"{confidence:.0%} CI: ({lower:.4f}, {upper:.4f})")
```

Based on these results, if the null hypothesis H0 is that p = 8, we reject this null at all three confidence levels tested and we can conclude that the mean is different from 8 at these confidence intervals. 8 falls outside of each of these confidence intervals

Statistically, these results are significant because we rejected the null and 8 hours fell outside of the confidence intervals. Practically, the difference between the observed mean and the null are not that different; they vary by about 20-ish minutes. This isn't that big a difference and whether it is practically significant would depend on what the result would be used for, but a scenario where that small of a difference mattered for sleep seems unlikely.
